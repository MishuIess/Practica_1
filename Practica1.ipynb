```python
# 1. Importación de librerías
import pandas as pd
import matplotlib.pyplot as plt
import random
from faker import Faker
fake = Faker()
```

```python
# 2. El archivo utilizado es un .csv de un dataset de kaglle link: https://www.kaggle.com/datasets/algozee/teenager-menthal-healy
# data = pd.read_csv("/content/drive/MyDrive/Practica_1/Teen_Mental_Health_Dataset.csv")
data = pd.read_csv("Teen_Mental_Health_Dataset.csv")
```

```python
print(data.head())
```

```python
# 3. Agregar cuatro columnas al dataframe
data = pd.DataFrame(data)
```

```python
#1. Nombre falso (Faker)
data["nombre"] = [fake.name() for _ in range(len(data))]

#2. Número de celular falso (Faker)
data["apellido"] = [fake.last_name() for _ in range(len(data))]

# 3. Email falso (Faker)
data["email"] = [fake.email() for _ in range(len(data))]

# 4. País falso (Faker)
data["pais"] = [fake.country() for _ in range(len(data))]
```

```python
print(data.head())
```

```python
# 4. Generar dos visualizaciones del dataframe
data["pais"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 países")
plt.xlabel("País")
plt.ylabel("Cantidad")

plt.xticks(rotation=45)
plt.show()
```

```python
data["sleep_hours"].plot(kind="hist", bins=100)

plt.title("Distribución de horas de sueño")
plt.xlabel("Horas de sueño")
plt.ylabel("Frecuencia")

plt.show()
```

```python
data.boxplot(column="stress_level")

plt.title("Distribución del estrés")
plt.show()
```
